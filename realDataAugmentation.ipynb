```python
import os
import glob
import numpy as np
import cv2
from PIL import Image
import pandas as pd
from collections import Counter

# Visualization
import seaborn as sns
import matplotlib.pyplot as plt

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn import metrics

# Deep Learning
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import tensorflow as tf
from torchvision import datasets, transforms, models

from torchvision.transforms.autoaugment import AutoAugment, AutoAugmentPolicy
```

```python
REAL_IMG_PATH = 'data/wiki/'
OUT_IMG_PATH = 'data/wiki_augmented/'
```

```python
import os
import random
from PIL import Image
import torchvision.transforms as transforms

# Define augmentation pipeline with randomness
augmentation = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    transforms.ToTensor()
])

os.makedirs(OUT_IMG_PATH, exist_ok=True)

# Number of augmented images per original
num_augmentations = 2  # 1 original + 2 augmented → 3x dataset

# Process each image
image_files_dirs = os.listdir(REAL_IMG_PATH)
print(image_files_dirs)
for sub_folder in image_files_dirs:
    sub_folder_path = os.path.join(REAL_IMG_PATH, sub_folder)
    
    # Skip if not a directory
    if not os.path.isdir(sub_folder_path):
        continue
        
    # Create corresponding output subdirectory
    os.makedirs(os.path.join(OUT_IMG_PATH, sub_folder), exist_ok=True)
    
    for img_file in os.listdir(sub_folder_path):
        # Skip if not an image file
        if not img_file.lower().endswith(('.png', '.jpg', '.jpeg')):
            continue
            
        img_path = os.path.join(sub_folder_path, img_file)
        image = Image.open(img_path).convert("RGB")  # Open image
        
        # Save original image to corresponding output subdirectory
        output_img_path = os.path.join(OUT_IMG_PATH, sub_folder, img_file)
        image.save(output_img_path)  
        
        # Generate augmented versions
        for i in range(num_augmentations):
            augmented_image = augmentation(image)
            save_path = os.path.join(OUT_IMG_PATH, sub_folder, f"{img_file.split('.')[0]}_aug{i}.jpg")
            
            # Convert tensor back to PIL and save
            transforms.ToPILImage()(augmented_image).save(save_path)

# print(f"Dataset expanded to {len(os.listdir(OUT_IMG_PATH))} images.")
```
